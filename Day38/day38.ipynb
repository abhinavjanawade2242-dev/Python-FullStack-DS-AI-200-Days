```python
corr=df[["CustomerID","UnitPrice"]].corr()
print(corr)
```

```python
# 5 Handle missing values
df["Description"]=df["Description"].fillna("No description")
```

```python
customer_df = df[df["CustomerID"].notna()].copy()

print("Rows available for customer analysis:", len(customer_df))
print("Missing IDs after filtering:", customer_df["CustomerID"].isnull().sum())
```

```python
print("Original dataset shape:", df.shape)
print("Number of duplicate rows:", df.duplicated().sum())
print(df[df.duplicated()].head())
```

```python
df_clean = df.drop_duplicates().copy()

print("Rows before cleaning:", len(df))
print("Rows after cleaning:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
print(df_clean.dtypes)
print("Zero or negative quantities:", (df_clean["Quantity"] <= 0).sum())
print("Zero or negative prices:", (df_clean["UnitPrice"] <= 0).sum())
```

```python
cancellations = df_clean["InvoiceNo"].astype(str).str.startswith("C")

print("Cancellation records:", cancellations.sum())
```
