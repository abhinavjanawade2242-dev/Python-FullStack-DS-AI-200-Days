```python
import pandas as pd

data = {
    "Name": [
        "Abhinav",
        " Rahul ",
        "Amit",
        "Rushi",
        "Omi",
        "Kiran",
        "Sanjay",
        "Vikas",
        "Abhinav",
        None
    ],

    "Age": [
        20,
        21,
        20,
        None,
        20,
        21,
        23,
        150,
        20,
        22
    ],

    "Department": [
        "Data Science",
        "Computer Science",
        "data science",
        "Computer Science",
        "Data Science",
        None,
        "Data Science",
        "Computer Science",
        "Data Science",
        "Computer Science"
    ],

    "Marks": [
        85,
        72,
        91,
        65,
        None,
        78,
        95,
        300,
        85,
        75
    ],

    "Attendance": [
        92,
        85,
        96,
        70,
        89,
        None,
        98,
        75,
        92,
        80
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 3 Understand the dataset
print(df.head())
print(df.tail())
print("Shape:",df.shape)
print("Columns:",df.columns)
print("DataType:",df.dtypes)
print("Dataset information:")
df.info()
print("Statistical Information:")
df.describe()
```

```python
#4 Missing values
print("missing values:",df.isnull().sum())
print("Missing value percentage:",(df.isnull().sum()/len(df))*100)
```

```python
#5 Find duplicated rows
print("Duplicated rows:",df.duplicated().sum())
print(df[df.duplicated()])
```

```python
# 6 Remove duplicated
df=df.drop_duplicates()
print("Duplicates:",df.duplicated().sum())
```

```python
# 7 Clean Text
df["Name"]=df["Name"].str.strip()
```

```python
# 8 Standardize categories
print(df["Department"])
df["Department"]=df["Department"].str.strip().str.title()
print(df["Department"].unique())
```

```python
# 9 Fill Missing values
df["Age"]=df["Age"].fillna(df["Age"].median())
df["Marks"]=df["Marks"].fillna(df["Marks"].median())
df["Attendance"]=df["Attendance"].fillna(df["Attendance"].median())
df["Department"]=df["Department"].fillna(df["Department"].mode()[0])
df=df.dropna(subset=["Name"])   #Drop row with no name
print(df)
```

```python
# 10 Detect Inavlid values
print(df[df["Age"]>100])
print(df[df["Marks"]>100])
```

```python
# 11 Fix Invalid values
df.loc[df["Age"]>100,"Age"]=None
df.loc[df["Marks"]>100,"Marks"]=None
df["Age"]=df["Age"].fillna(df["Age"].median())
df["Marks"]=df["Marks"].fillna(df["Marks"].median())
```

```python
# 12 Detect outlier
def outlier(df,columns):
    Q1=df[columns].quantile(0.25)
    Q3=df[columns].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    outliers=[
        df[
            (df[columns]<lower_bound) |
            (df[columns]>upper_bound)
        ]
    ]
    return outliers
```

```python
print("Mark outliers:",outlier(df,"Marks"))
print("Attendance outliers:",outlier(df,"Attendance"))
```

```python
# 13 After cleaning validate ranges
print("Invalid Age:",df[(df["Age"]<15) | (df["Age"]>30)])
print("Invalid Marks:",df[(df["Marks"]<0 | (df["Marks"]>100))])
print("Invalid Attendance:",df[(df["Attendance"]<0) | (df["Attendance"]>100)])
```

```python
# 14 review the dataset
print(df)
print("Shape:",df.shape)
print("Missing Values:\n",df.isnull().sum())
print("Duplicate values:\n",df.duplicated().sum())
print("DataType:\n",df.dtypes)
print("Statistics:\n",df.describe())
```
