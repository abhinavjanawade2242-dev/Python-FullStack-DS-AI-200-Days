```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\Abhinav\Python-FullStack-DS-AI-200-Days\Day36\Data\students.csv")
df.head(5)
```

```python
# 3 Understand dataset
print("Shape:",df.shape)
print("Columns:",df.columns)
print("DataTypes:",df.dtypes)
print("Information:",df.info())
```

```python
# 4 Missing values
print(df.isnull().sum())
df=df.dropna(subset=["internet"])
df["attendance"]=df["attendance"].fillna(df["attendance"].median())
df["G2"]=df["G2"].fillna(df["G2"].median())
print(df.isnull().sum())
```

```python
# 5 Check duplicates
print(df.duplicated().sum())
df=df.drop_duplicates()
print(df.duplicated().sum())
```

```python
# 6 statistical summary
print(df.describe())
```

```python
numerical_columns=df.select_dtypes(include=np.number).columns
print(numerical_columns)
```

```python
categorical_columns=df.select_dtypes(include=["str","object"]).columns
print(categorical_columns)
```

```python
# 10 Visualization-Distribution
sns.histplot(
    df["age"],
    kde=True
)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of students")
plt.show()
```

```python
# 11 BoxPlot
sns.boxplot(x=df["age"])
plt.title("Age Distribution")
plt.show()
```

```python
# 12 Count Plot
sns.countplot(
    data=df,
    x="sex"
)
plt.title("Students by gender")
plt.xlabel("Sex")
plt.ylabel("No. of students")
plt.show()
```

```python
# 13 Compare two variables
sns.boxplot(
    data=df,
    x="sex",
    y="G3"
)
plt.title("sex v/s G3")
plt.show()
```

```python
# 14 scatterplot
sns.scatterplot(
    data=df,
    x="sex",
    y="G3"
)
plt.title("Sex v/s G3")
plt.show()
```

```python
# 15 Correlation
correlation=df[numerical_columns].corr()
print(correlation)
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)
plt.title("HeatMap")
plt.show()
```

```python
# 16 Strong correlation
print(correlation["G3"].sort_values(ascending=False))
```

```python
# 17 groupby()
# 18 Outlier
Q1=df["G3"].quantile(0.25)
Q3=df["G3"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
outlier=df[(df["G3"]<lower_bound) | (df["G3"]>upper_bound)]
print("Outliers:",outlier)
```
